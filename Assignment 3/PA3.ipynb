## Task 1: Types of Triangles

```python
# A. 
side1=int(input("Please enter the length of the first side of your triangle: "))
side2=int(input("Now enter the length of the second side: "))
side3=int(input("Finally, enter the length of the third side: "))
# B. 
# B. 
if side1==side2==side3:
    print('You have an equilateral triangle!')
elif side1!=side2 and side1==side3 or side2!=side3 and side1==side2 or side1!=side3 and side3==side2:
    print('You have a non-equilateral isosceles triangle!')
else:
    print('You have a scalene triangle!')
```

## Task 2: Student Grades and At-Risk Status

```python
# A.
student_grades=[]
lowgrades=[]
while len(student_grades)<=5:
    newGrade=float(input('Enter your grade here: '))
    student_grades.append(newGrade)
    if newGrade<=65:
        lowgrades.append(newGrade)

```

```python
# B. 
## Average:
print(sum(student_grades)/len(student_grades))
## Minimum:
print(min(student_grades))
## Maximum:
print(max(student_grades))
```

```python
# C. 
print(len(lowgrades),' students are at risk in the course.')
```

```python
# Just in case you need to blank the list to test it again...
del(student_grades)
```

## Task 3: Ranges and Multiples

```python
# A. 
mult_of_2=[]
numbers2=0
while numbers2<=100:
    numbers2+=1
    if numbers2%2==0:
        mult_of_2.append(numbers2)
print("Multiples of 2: ",mult_of_2)
```

```python
# B. 
mult_of_3=[]
numbers3=0
while numbers3<=100:
    numbers3+=1
    if numbers3%3==0:
        mult_of_3.append(numbers3)
print("Multiples of 3: ",mult_of_3)
```

```python
# C. 
mult_of_both=[]
numbersBoth=0
while numbersBoth<=100:
    numbersBoth+=1
    if numbersBoth%2==0 and numbersBoth%3==0:
        mult_of_both.append(numbersBoth)
print("Multiples of both 2 and 3: ",mult_of_both)
```

```python
# D.
mult_of_either=[]
numbersEither=0
while numbersEither<=100:
    numbersEither+=1
    if numbersEither%2==0 or numbersEither%3==0:
        mult_of_either.append(numbersEither)
print("Multiples of either 2 or 3: ",mult_of_either)
```

```python
# Here's an alternate version where everything is combined into one list!
mult_of_2_2=[]
mult_of_3_2=[]
mult_of_both_2=[]
mult_of_either_2=[]
number=0
while number<=100:
    number+=1
    if number%2==0:
        mult_of_2_2.append(number)
    if number%3==0:
        mult_of_3_2.append(number)
    if number%3==0 and number%2==0:
        mult_of_both_2.append(number)
    if number%3==0 or number%2==0:
        mult_of_either_2.append(number)
print("Multiples of 2: ",mult_of_2_2)
print("Multiples of 3: ",mult_of_3_2)
print("Multiples of both 2 and 3: ",mult_of_both_2)
print("Multiples of either 2 or 3: ",mult_of_either_2)
```

## Task Four: Shopping List

```python
# A. 
prompt="Create your shopping list!"
shopping_list=[]
end="Done"

while len(shopping_list)<7:
    item=input('What do you need to purchase?')
    if item in shopping_list:
        print("You've already added that item!")
        continue
    elif item.lower()==end.lower():
        break
    else: 
        shopping_list.append(item)
print("That's all I need!")
```

```python
# B. 
print("You have",len(shopping_list),"items on your shopping list.")
```

```python
# C. 
shopping_list.reverse()
print(shopping_list)
```

```python
# D. 
shopping_list.sort()
print(shopping_list)
```
