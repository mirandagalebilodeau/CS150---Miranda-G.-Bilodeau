```python
import random

#Creates a 4 digit secret code using the randrange function to assign nonduplicate numbers to a list called secret_code
wins=0
losses=0
active=True

secret_code=[]
while len(secret_code)<4:
    code=random.randrange(0,10)
    if code not in secret_code:
        secret_code.append(code)
        

#number of wins and losses
wins=0
losses=0

#number of guesses the user has
attempts=10 

#number of correct numbers in position var
correctposition=0

#number of correct numbers not in correct positions var
correctnum=0

print('Welcome to Mastermind! You are supposed to figure out the secret 4 digit code. Each guess you will enter 4 numbers, starting from left to right. You have 10 attempts.')

#While loop that keeps the game running as long as the guess_code is not correct and user has attempts left
active=True
while active: 
#Take user input to get 4 numbers for their guess
    try:
        guess=input('Enter your 4 digit guess:')
#Take the 4 number guess and create it into a list to compare to secret_code
        guess_code=[int(digit) for digit in guess]
    
#For loop that checks the guess_code var to the secret_code var to see if they match
        if len(guess)==4:
            for i in range(4):
                if guess_code[i]==secret_code[i]:
                    correctposition+=1
                elif guess_code[i] in secret_code:
                    correctnum+=1
            print(f'Correct numbers in position: {correctposition} and correct numbers in wrong position: {correctnum}')
            attempts-=1
            print('You Have',attempts,'Attempts Remaining')
            correctposition=0
            correctnum=0
        else:
            print('Your guess must be four digits long!')
    except ValueError:
        print('Your guess must be an integer!')
        continue
    except IndexError:
        print('Your guess must be four digits long!')
        continue
#If statements that check if the user has won or lost the game
    if attempts>=1 and guess_code!=secret_code:
            continue
    elif guess_code==secret_code:
        wins+=1
        print(f'You have won! Your total score is now {wins} win(s) and {losses} loss(es).')
    else:
        losses+=1
        print(f'You have lost. Your total score is now {wins} win(s) and {losses} loss(es). Would you like to try again?')
        play=input(' Would you like to play again? Yes or No:')
#if statement that will restart the game if user has entered Yes
    if play.lower()=='yes':
        active=True
        attempts=10
        secret_code=[]
        while len(secret_code)<4:
            code=random.randrange(0,10)
            if code not in secret_code:
                secret_code.append(code)
    else:
        print('Thank you for playing! Have a lovely day!')
        active=False
```
