## Task 1: Weekend Activities as Sets

```python
actSat=set()
active=True
print('What activities are you planning to do this Saturday?')
while active:
    activity=input()
    if activity.lower()=='done':
        print('All done!')
        active=False
    else:
        actSat.add(activity)
    
```

```python
actSun=set()
active=True
print('What activities are you planning to do this Sunday?')
while active:
    activity=input()
    if activity.lower()=='done':
        print('All done!')
        active=False
    else:
         actSun.add(activity)
   
```

```python
# C.
print(actSat)
print(actSun)
# D. 
bothdays=[]
for activity in actSat:
    if activity in actSun:
      bothdays.append(activity)
print(bothdays)
# E. 
eitherday=[]
eitherday.extend(actSat)
eitherday.extend(actSun)
print(eitherday)
```

## Task 2: Triangle Calculations

```python
def triangle_solve(leg1,leg2):
    hyp=((leg1**2)+(leg2**2))**0.5
    per=leg1+leg2+hyp
    area=0.5*leg1*leg2
    stats=(hyp,per,area)
    print(stats)
```

```python
triangle_solve(3,4)
```

## Task 3: Calculator Class

```python
class Calculator():
    def __init__(self,a,b):
        self.a=a
        self.b=b
        self.total_calculations=0
        self.prev_result=0
    def add(self):
        self.prev_result-=self.prev_result
        self.prev_result+=self.a+self.b
        self.total_calculations+=1
        return self.a+self.b
    def subtract(self):
        self.prev_result-=self.prev_result
        self.prev_result+=self.a-self.b
        self.total_calculations+=1
        return self.a-self.b
    def multiply(self):
        self.prev_result-=self.prev_result
        self.prev_result+=self.a*self.b
        self.total_calculations+=1
        return self.a*self.b
    def divide(self):
        self.prev_result-=self.prev_result
        self.prev_result+=self.a/self.b
        self.total_calculations+=1
        return self.a/self.b
    def power(self):
        self.prev_result-=self.prev_result
        self.prev_result+=self.a**self.b
        self.total_calculations+=1
        return self.a**self.b
    def info(self):
        print(f'Previous result: {self.prev_result}')
        print(f'Total calculations made: {self.total_calculations}')
        
```
