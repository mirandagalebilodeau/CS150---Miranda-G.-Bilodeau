## Task 1: Input Positive Integer and Factors

```python
# A. 
num=int(input('Please enter a positive integer here:'))
for f in range(1,num+1):
    if num>0:
        if num%f==0:
            print(f'{f} is a factor of {num}')

        
    
```

```python
# B. 
num2=int(input('Please enter a positive integer here:'))
primecount=[]
for f in range(1,num2):
   if num2%f!=0:
       primecount.append(f)
   else:
       continue
if len(primecount)==len(list(range(1,num2)))-1:
    print('This is a prime number!')
else: 
    print('This is not a prime number.')
```

## Task 2: Language Translator

```python
# This is a quick little English-Nahuatl dictionary focused on animals! All words were sourced from this site: https://www.native-languages.org/nahuatl_animals.htm
nahua_en={'dog':'chichi' , 'spider':'tocatl' , 'snake':'coatl' , 'monkey':'ozomahtli' , 'bee':'xicohuitl' , 'fish':'michin' , 'turtle':'ayotl' , 'bat':'quimichpatlan' , 'parrot':'cochotl' , 'mouse':'quimichin' , 'owl':'tecolotl'}
active=True
while active == True:
    query=input('Enter a word to translate from English to Nahuatl:')
    if query:
        if query in nahua_en.keys():
            print(f'The Nahuatl word for {query} is {nahua_en[query]}')
        else:
            print(f'The Nahuatl word for {query} is not in this dictionary.')
    else:
        print('Thank you for using this dictionary! Bye for now!')
        active=False
        
```

## Task 3: Temperature Summaries


```python
## I'm using the data from this website: https://www.weather.gov/lmk/clisdf
louTemp={'January':[43.6,27.8],'February':[48.3,30.7],'March':[58.1,38.6],'April':[69.6,48.5],'May':[77.8,58.7],'June':[85.7,67.2],'July':[89.0,70.8],'August':[88.4,69.5],'September':[82.2,61.9],'October':[70.5,50.1],'November':[57.6,39.4],'December':[47.2,32.1]}
print('Curious about the average temperature for each month in Louisville, KY? Look no further!')

active=True
while active==True:
    query=input('Enter a month here to see the average high and low temperature:').capitalize()
    if query!='Quit':
        if query in louTemp.keys():
            print(f'In Louisville, the average high temperature in {query} is {louTemp[query][0]} degrees Fahrenheit, and the average low temperature is {louTemp[query][1]} degrees Fahrenheit.')
        else:
            print('That is not a valid month!')
    else:
        active=False





```

## Task 4: Inventory Dictionary and Prices

```python
inventory = {'hammer':[15, 12.49], 'wrench':[20, 7.99], 'screwdriver': [40, 3.99], 'nailgun':[10, 36.59], 'padlock':[25, 5.99], 'roll_of_duct_tape': [30, 4.95]} 
```

```python
print('Welcome to the hardware store! Enter "1" to see the price of an item and the number of items available, "2" to see which items are priced below a custom maximum budget, "3" to see the total quantity of items available, or "4" to see the total price of all stock.')

menu_option=int(input('Enter your number here:'))
if menu_option==1:
    active=True
    while active==True:
        item=input('Which item would you like more information about? Enter "quit" to end the loop').lower()
        if item!='quit': 
            if item in inventory.keys():
                print(f' Each {item} is ${inventory[item][1]}, and we have {inventory[item][0]} available.')
            else: 
                print(f' Apologies, we do not have any in stock right now!')
        else:
            print('Thank you for your time!')
            active=False
elif menu_option==2:
    max_budget=float(input('Please enter your maximum budget here! No dollar signs, please!'))
    print(f' Your maximum budget is ${max_budget}.')
    for item in inventory:
        if inventory[item][1]<=max_budget:
            print(f' Each {item} is ${inventory[item][1]}. It falls within your budget!')
        else:
            continue
elif menu_option==3:
    print(f' In total, we have {15+20+40+10+25+30} items in stock.')
elif menu_option==4:
    print(f' The total cost of every item in our inventory together is {(15*12.49) + (20*7.99) + (40*3.99) + (10*36.59) + (25*5.99) + (30*4.95)} dollars.')
else:
    print(f' Apologies, that is not an option!')

```
